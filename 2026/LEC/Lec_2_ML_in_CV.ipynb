<font size=4.5pt  face = 'georgia' style='Line-height :2'><p style='text-align: left;'>   
    <b>The nonlinear regression model </b> is the basic part of the neural network. <br>
    The problem can be given as
    $$\hat{y}_i = 𝑓(\sum_{𝑗=0}^{𝑀} 𝑤_𝑗 𝑥_{𝑖𝑗} )= f(W^TX_i), $$
    where <ul>
        <li> $\hat{y}_i$ is the some deicisin;
        <li> $W = \{𝑤_𝑗\}_{j=0}^M$ is the the set of the model parameters (need to be estimated, including weights $\{𝑤_𝑗\}_{j=1}^M$ and bias $w_0$ );
        <li>$f(\cdot)$ is the activation function - any arbitrary function (could be with non-linear behavior); 
        <li> $X_i = \{x_{ij}\}_{j=1}^{M}$ is the some input (could be tensor (flatten to vector) in general).  
    </ul>  
  <u>The meaning of the activation function is to describe some non-linear relations for the input-output.</u><br>   <font size=4.5pt  face = 'georgia' style='Line-height :2'><p style='text-align: left;'>   
We can formulate the problem of nonlinear regression learning as:
    $$ err_i=L_i = L(y_i,\hat y_i) =  L\big(y_i , f(\sum_{j=0}^M w_jx_{ij})\big)=L\big(y_i , f(WX_i)\big) \to min, $$
where <ul>
    <li> $y_i$ is the our answer.
    <li> $L$ is the loss function - analouge of error.  </ul>
<font size=4.5pt  face = 'georgia' style='Line-height :2'><p style='text-align: left;'>       
    In the particular case $$L_i = L\big(y_i - f(WX_i)\big)$$  <br>
    It is valuable to note that frequently we can call  $𝐿 $ - the likelihood. <ul>
    Frequently, it is accepted and valuable to replace the $ 𝐿 $ with it logarithm. </ul>
    




<font size=4.5pt  face = 'georgia' style='Line-height :3'><div style='text-align: justify; padding:3px; margin-top:-1px; margin-bottom:-1px'> 
    The problem of training non-linear regrerssors can be written as 
    $$ L_i = L(y_i,W,X_i)=L\big(y_i -f(W^TX_i)\big)\to min, $$ 
where <ul>
    <li> $L_i $ is the error of training, ( the loss function);
    <li> $f(W^TX_i)$ is the some non-linear result of the network or its layer
        <ul>
            <li> $W$ is the weights matrix;
            <li> $X_i$ is the layer input;
            <li> $f(\cdot)$ is the activation function. 
        </ul>   
    </ul>
<font size=4.5pt  face = 'georgia' style='Line-height :3'><div style='text-align: justify; padding:3px; margin-top:-1px; margin-bottom:-1px'>     
Thus the our task is to minimize the error i.e. minimize the loss function value.    
    <br><br>
In other words, we can proposed that for optimal weights $W$ the value of $L_i$ will be sufficiently small. 
    <br><br>
Hence the task is to select the $W$ values optimally.
<font size=4.5pt  face = 'georgia' style='Line-height :3'><div style='text-align: justify; padding:3px; margin-top:-1px; margin-bottom:-1px'> 
The training problem could be solved iteratively. This approach is also called <b>The Gradient Descent Method</b>
    $$
    W^t=W^{t-1} -\eta \nabla L(W^{t-1},X_i,y_i),
    $$
where $\eta$ is the learning rate.
    <br>  
In the prectice we will calculate the weights update by a set of samples (set of $i$). In this case we need to rewrite this equation as
     $$ W^t=W^{t-1} -\eta \frac{1}{N}\sum_{i=0}^{N-1}\nabla L(W^{t-1},X_i,y_i),$$
     where <ul>
         <li> N is the number of samples (i.e. <b>Batch</b>).
         <li> $\frac{1}{N}\sum_{i=0}^{N-1} L(W,X_i,y_i)$ is the so-called <b>cost function</b>.  
     </ul>
If we are taking all samples in the batch randomly, than we have <b>Stochastic Gradient Descent</b> (<b>SGD</b>).<br>    
<b>The backward propagation method Idea:</b><ul>
        <li>calculation of how the obtained error change for each perceptron in propagation in the direction from output to input.<br>
        <li>If we understand this, we can update each weight on the optimal value.<br>
        <li>The method is optimization scheme of gradient descent for graph-based function, imply two stages:<ul>
            <li>1. forward propagation stage on which outputs are calculated.<br>
             <li>2. backward propagation stage on which the error is propagated in back direction and weights updated by error obtained values. 
            </ul> </ul>
<b>The backpropagation rules</b> as:    
$$\Delta W_l \leftarrow \frac{1}{N} \sum_{i=0}^{N-1}  {\nabla_{W_l}} L(y_i, \hat y_i) = \frac{1}{N} \sum_{i=0}^{N-1} \left(e_{l} \odot \frac{\partial f(z_{l})}{\partial z_l} \right) a_{l-1}^T  $$    
and
$$W_l^{t} = W_l^{t-1} - \eta \frac{1}{N} \sum_{i=0}^{N-1} \nabla_{W_l}L(y_i,\hat{y}_i) $$

    The main advantage of using deep neural network is automatic feature extraction. <br>
        The extraction making sequentially layer by layer.
            <li> the weight (kernel) the same for each part of the picture (weights sharing).
            <li> the results of such convolution weakly depends on the position, size, and location of each object or pattern (or each feature) location, scale, and rotation invariant for convolution result.  

<font size=4.5pt  face = 'georgia' style='Line-height :3'><div style='text-align: justify; padding:5px; margin-top:-5px; margin-bottom:-5px'>
It can be assumed how the simplest CNN (for classification) look-like: 
    $$ \hat{y} = f_2(w\cdot \vec{vec}[\text{pooling}({f}_1(k\ast x))]),$$ 
    where <ul>
    <li> $\hat{y}$ is the classification output.
    <li> $w$ is the vector of  classification weights. 
    <li> $\text{pooling}$ is the operation for dimension reduction (subsampling).  
    <li> $\text{vec}$ is the operation of matrix vectorization (reshape, revel).     
    <li> $f_1$ and $f_2$ are the <b>activation functions </b>.<ul>
    <li> $\ast$ is the operation of convolution.    
        <li>For input layers $f_1$ is the frequently any kind of <b>Rectified Linear Unit (ReLU)</b> functions (simplest $\textbf{relu}(x) = \text{max}(0,x)$ ).
        <li>For output classification layers $f_2$ is the frequently any kind of probability producing functions (often $\textbf{softmax}(x_i) = \text{exp}(x_i)/\sum_{i}\text{exp}(x_i) $   ).      
    </ul> 
    </ul>    
<em>Note</em> In the basic idea the sequence of operation in the layer is $\text{conv} \to \text{activation function} \to \text{pooling}$

<font size=4.5pt  face = 'georgia' style='Line-height :2'><div style='text-align: justify; padding:35px'>
The principle of the convolution work for 1 channel is the following operation
<span style='text-align: justify; padding:35px; font-size:x-large' >    
$$ r[i,j] = (k\ast x)[ij] = \sum_{a=0}^{k_h-1}\sum_{b=0}^{k_w-1}k[a,b]x[i+a,j+b], $$</span>  
where $r$ is the convolution output for convolution of input image $x$ and kernel $k$ with size $k_h\times k_w$. 
<br> 
 The convolution operation is equal to the sliding by the filter through the image as it shown in the example: 
    

<!-- ![](https://cdn-media-1.freecodecamp.org/images/gb08-2i83P5wPzs3SL-vosNb6Iur5kb5ZH43) -->

<font size=4.5pt  face = 'georgia' style='Line-height :2'><div style='text-align: justify; padding:35px'>
So we can actually rewrite convolution using pointwise (or element-wise, Hadamard) product $\odot$ as
```python
r[0,0] = np.sum(x[:3,:3]  * k) 
r[1,0] = np.sum(x[1:4,:3] * k)
r[2,0] = np.sum(x[2:5,:3] * k)
r[0,1] = np.sum(x[:3,1:4] * k) 
r[0,2] = np.sum(x[:3,2:5] * k) 
```

$$ r[ij] = \sum_{ij}{x[i:i+k_h,j:j+k_w]\odot k}$$

<font size=4.5pt  face = 'georgia' style='Line-height :2'><div style='text-align: justify; padding:35px'>
    Beside the simple convolution (<b>direct loop convolution</b>) representation as it shown above, it can be proposed several types of acceleration the convolution operation:<ol>
<li> <b>direct loop convolution</b>.
<li> <b> Image to column convolution (im2col)</b>.
<li> <b> kernel to row convolution (kn2row)</b>.        
<li> <b>Convolution though the Fast Fourier Transform (FFT)</b>:<ul>
    $$r = \mathcal{F}^{-1}( \mathcal{F}(k)\cdot\mathcal{F}(x)),$$
where $\mathcal{F}$ and $\mathcal{F}^{-1}$ are the direct and inverse Fast Fourier Transforms; and in the case of the $k$ the kernel is zero-padded to the shape of image.</ul>
<li> <b>Fast Winograd algorithm (Winograd)</b>: method is based on the dividing the input image to a set of small blacks and simplification the complexity for convolution for each of the blocks.

As a rule a set of convolution kernels simultaneously applied for a input tensor (set of feature maps)<br> for obtain one output feature map.<br>
Thus, in fact you obtain 3d-convolution which is a sum of 2d-convolution with a corresponding set of kernels.<br> 
$$ r[i,j] = (k\ast x)[ij] = \sum_{c=0}^{C-1} \big[ \sum_{a=0}^{k_h-1}\sum_{b=0}^{k_w-1}k^с[a,b]x^c[i+a,j +b]\big], $$
where $C$ is the number of input feature maps and $k^с$ is the kernel for $c$-th feature map.  <br> 
For obtaining several output feature maps we need  several 3-d sets of kernels.
